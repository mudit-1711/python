```python
em['text']=em['text'].apply(lambda x : x.lower())
```

```python
import string
def rem_pun(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
em['text']=em['text'].apply(rem_pun)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

em['text'] = em['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

em['text'] = em['text'].apply(remove_emojis)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
print(len(stop_words))
stop_words
```

```python
em.loc[1]['text']
```

```python
def remove(txt):
    words=word_tokenize(txt)
    cleaned_text=[]
    for i in words:
        if not i in stop_words:
            cleaned_text.append(i)
    return ' '.join(cleaned_text)            
```

```python
em['text']=em['text'].apply(remove)
```
