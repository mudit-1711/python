```python
unique_em=em['emotion'].unique()
```

```python
emotions_num={}
i=0
for e in unique_em:
    emotions_num[e]=i
    i+=1
em['emotion']=em['emotion'].map(emotions_num)    
```

```python
em['text']=em['text'].apply(lambda x : x.lower())
```

```python
import string
def rem_pun(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
em['text']=em['text'].apply(rem_pun)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

em['text'] = em['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

em['text'] = em['text'].apply(remove_emojis)
```
