```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('netflix_titles.csv')
```

```python
df.info()
```

```python
df = df.dropna(subset=['show_id', 'type', 'title', 'director', 'cast', 'country', 'date_added', 'release_year', 'rating', 'duration', 'listed_in', 'description'])
```

```python
df.head()
```

```python
df.info()
```

```python
type_count = df['type'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_count.index,type_count.values,color=['pink','blue'])
plt.title('Number of movies vs TV show on netflix')
plt.xlabel('type')
plt.ylabel('count')
plt.tight_layout()
plt.savefig('movie_vs_tvshow.png')
plt.show()
```

```python
rating_count=df['rating'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(rating_count,labels=rating_count.index,autopct='%1.1f%%',startangle=90)
plt.title('percent of content rating')
plt.tight_layout()
plt.savefig('content_rating_pie.png')
plt.show()
```

```python
movies = df[df['type'] == 'Movie']

plt.figure(figsize=(8,6))
plt.hist(movies['release_year'], bins=20, edgecolor='black',color='purple')

plt.title('Distribution of Movie Release Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies')

plt.tight_layout()
plt.savefig('movie relase year histogram.png')
plt.show()
```

```python
movies = df[df['type'] == 'Movie'].copy()
movies['duration'] = movies['duration'].str.replace(' min', '').astype(int)

plt.figure(figsize=(8,6))
plt.hist(movies['duration'], bins=20, edgecolor='black', color='purple')

plt.title('Distribution of Movie Duration')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')

plt.tight_layout()
plt.savefig('movie_duration_histogram.png')
plt.show()
```
