```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X,y_true= make_blobs(n_samples=500,centers=3,cluster_std=0.6,random_state=42)
```

```python
X
```

```python
y_true
```

```python
df=pd.DataFrame(X,columns=['f1','f2'])
```

```python
df
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
```

```python
inertia=[]
k_range=range(1,11)
```

```python
for k in k_range:
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit(x_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(k_range,inertia,marker='o',color='seagreen',linestyle='--')
```

```python
kmeans_final=KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels=kmeans_final.fit_predict(x_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x=df['f1'],
                y=df['f2'],
                hue=df['cluster'],
                palette='viridis')
```

```python
from sklearn.cluster import  DBSCAN
from sklearn.datasets import  make_moons
```

```python
X,y_true = make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
from sklearn.cluster import  KMeans
```

```python
df=pd.DataFrame(X,columns=['f1','f2'])
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
kmeans_label = kmeans.fit_predict(x_scaled)
```

```python
df['kmeans_cluster'] = kmeans_label
```

```python
sns.scatterplot(x=df['f1'],
                y=df['f2'],
                hue=df['kmeans_cluster'],
                palette='viridis')
```

```python
sns.scatterplot(x=df['f1'],
                y=df['f2'],
                hue=df['kmeans_cluster'],
                palette='tab10')
```

```python
dbscan = DBSCAN(eps=0.3, min_samples=5)
dbscan_labels = dbscan.fit_predict(x_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x=df['f1'], y=df['f2'], hue=df['dbscan_cluster'], palette='tab10')
```
